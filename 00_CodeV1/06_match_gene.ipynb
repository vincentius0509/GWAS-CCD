```python
# gtf_file = "/home/yhSong/data/02_ByProject/DogAgingProject/GeneticData/analysis/01_GWAS/6.Canis_lupus_familiaris.CanFam3.1.104.gtf.gz"
gtf_file = "../../Canis_lupus_familiaris.CanFam3.1.104.gtf.gz"

gtf_columns = ['seqname', 'source', 'feature', 'start', 'end', 'score', 'strand', 'frame', 'attribute']
gtf_data = pd.read_csv(gtf_file, sep='\t', comment='#', names=gtf_columns)
```

```python
# GTF 파일에서 gene ID 추출 함수
def extract_gene_id(attribute):
    for entry in attribute.split(';'):
        if entry.strip().startswith('gene_id'):
            return entry.split('"')[1]
    return None
```

```python
# gene ID 컬럼 추가
gtf_data['gene_id'] = gtf_data['attribute'].apply(extract_gene_id)

gtf_data.head()
```

```python
# gtf_data를 복사하여 gtf_data_modified에 저장
gtf_data_modified = gtf_data.copy()

# seqname의 값을 문자형으로 변환
gtf_data_modified['seqname'] = gtf_data_modified['seqname'].astype(str)

# X를 39로, 39를 X로 매핑할 수 있도록 변경
gtf_data_modified['seqname'] = gtf_data_modified['seqname'].replace({'X': '39', '39': 'X'})


snp_data['CHR'] = snp_data['CHR'].astype(str)
```

```python
# 결과 저장을 위한 데이터프레임 초기화
matched_genes = []
chr_genes_combined = pd.DataFrame() # 모든 chr_genes 데이터를 저장할 데이터프레임

# SNP와 gene 매칭
for idx, snp in snp_data.iterrows():
    # SNP의 염색체 번호와 일치하는 gene 데이터 필터링
    chr_genes = gtf_data_modified[(gtf_data_modified['seqname'] == snp['CHR']) & (gtf_data_modified['feature'] == 'gene')]
    
    # chr_genes 데이터를 chr_genes_combined에 추가
    chr_genes_combined = pd.concat([chr_genes_combined, chr_genes], ignore_index=True)
    
    # SNP의 위치(POS)가 gene의 시작과 끝 범위 안에 포함되는지 확인
    match = chr_genes[(chr_genes['start'] <= snp['POS']) & (chr_genes['end'] >= snp['POS'])]
    
    if not match.empty:
        # 매칭된 gene 데이터를 결과에 추가
        for _, gene in match.iterrows():
            matched_genes.append({
                'CHR': snp['CHR'], 
                'SNP': snp['SNP'], 
                'POS': snp['POS'], 
                'gene_id': gene['gene_id']
            })

# 결과 데이터프레임 생성
matched_genes_df = pd.DataFrame(matched_genes)
```

```python
# SNP와 매칭된 유전자 위치를 찾기 위한 리스트 초기화
matched_features = []

# SNP와 각 기능 매칭
for idx, snp in snp_data.iterrows():
    # SNP의 염색체 번호와 일치하는 모든 기능 데이터 필터링
    chr_features = gtf_data_modified[(gtf_data_modified['seqname'] == snp['CHR'])]
    
    # SNP의 위치(POS)가 기능의 시작과 끝 범위 안에 포함되는지 확인
    match = chr_features[(chr_features['start'] <= snp['POS']) & (chr_features['end'] >= snp['POS'])]
    
    if not match.empty:
        # 매칭된 기능 데이터를 결과에 추가
        for _, feature in match.iterrows():
            matched_features.append({
                'CHR': snp['CHR'],
                'SNP': snp['SNP'],
                'POS': snp['POS'],
                'gene_id': feature['gene_id'],
                'feature_type': feature['feature'],  # 기능 유형 추가 (예: exon, CDS 등)
                'feature_start': feature['start'],
                'feature_end': feature['end']
            })

# 결과 데이터프레임 생성
matched_features_df = pd.DataFrame(matched_features)
```

```python
# 각 feature_type별 고유 gene_id의 수를 계산하는 함수
def count_unique_genes_by_feature(feature_name):
    # 특정 feature_name에 해당하는 행만 필터링
    feature_df = matched_features_df[matched_features_df['feature_type'] == feature_name]
    # 고유한 gene_id의 수 계산
    return feature_df['gene_id'].nunique()

# 각 feature_type에 대해 고유 gene_id 수 계산
gene_count = count_unique_genes_by_feature('gene')
transcript_count = count_unique_genes_by_feature('transcript')
exon_count = count_unique_genes_by_feature('exon')
cds_count = count_unique_genes_by_feature('CDS')
five_prime_utr_count = count_unique_genes_by_feature('five_prime_utr')
three_prime_utr_count = count_unique_genes_by_feature('three_prime_utr')

# 결과 출력
print(f"feature_type이 'gene'인 유전자의 수: {gene_count}")
print(f"feature_type이 'transcript'인 유전자의 수: {transcript_count}")
print(f"feature_type이 'exon'인 유전자의 수: {exon_count}")
print(f"feature_type이 'CDS'인 유전자의 수: {cds_count}")
print(f"feature_type이 'five_prime_utr'인 유전자의 수: {five_prime_utr_count}")
print(f"feature_type이 'three_prime_utr'인 유전자의 수: {three_prime_utr_count}")
```
